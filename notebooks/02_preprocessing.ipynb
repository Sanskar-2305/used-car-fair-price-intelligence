```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/cars_cleaned.csv")

print(f"Rows: {len(df):,}")
print(f"Columns: {df.shape[1]}")

df.head()
```

```python
categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    print(f"\n===== {col} =====")
    print("Unique values:", df[col].nunique())
    print("Missing:", df[col].isna().sum())
    print(df[col].dropna().value_counts().head(10))
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    print(f"\n===== {col} =====")
    print("Missing:", df[col].isna().sum())
    print("Unique values:", df[col].nunique())
    print(df[col].value_counts(dropna=False).head(10))
```

```python
for col in ["engine", "transmission", "drivetrain", "fuel_type", "mpg"]:
    print(f"\n===== {col} =====")
    print("Unique:", df[col].nunique())
    print("Missing:", df[col].isna().sum())
    print(df[col].dropna().value_counts().head(15))
```

```python
for col in ["accidents_or_damage", "one_owner", "personal_use_only"]:
    print(f"\n===== {col} =====")
    print(df[col].value_counts(dropna=False))
```

```python
df["vehicle_age"] = 2024 - df["year"]

df["vehicle_age"].describe()
```

```python
df["has_accident"] = df["accidents_or_damage"]
df["is_one_owner"] = df["one_owner"]
df["is_personal_use"] = df["personal_use_only"]
```

```python
df["engine_liters"] = (
    df["engine"]
    .str.extract(r"(\d+(?:\.\d+)?)L", expand=False)
    .astype(float)
)
```

```python
df[["engine", "engine_liters"]].head(15)
```

```python
mpg_parts = df["mpg"].str.split("-", expand=True)

df["mpg_high"] = pd.to_numeric(mpg_parts[0], errors="coerce")
df["mpg_low"] = pd.to_numeric(mpg_parts[1], errors="coerce")
```

```python
df[["mpg", "mpg_high", "mpg_low"]].head(20)
```

```python
df["mpg_avg"] = df[["mpg_high", "mpg_low"]].mean(axis=1)
```

```python
df[["mpg", "mpg_high", "mpg_low", "mpg_avg"]].head(20)
```

```python
df["mpg_avg"].describe()
```

```python
print("Minimum MPG:", df["mpg_avg"].min())
print("Maximum MPG:", df["mpg_avg"].max())
print("Missing MPG:", df["mpg_avg"].isna().sum())
```

```python
df.loc[
    (df["mpg_avg"] <= 0) | (df["mpg_avg"] > 80),
    ["manufacturer", "model", "year", "mpg", "mpg_high", "mpg_low", "mpg_avg"]
].head(30)
```

```python
print("MPG <= 0:", (df["mpg_avg"] <= 0).sum())
print("MPG > 80:", (df["mpg_avg"] > 80).sum())
```

```python
df.loc[
    (df["mpg_avg"] <= 0) | (df["mpg_avg"] > 80),
    ["manufacturer", "model", "year", "fuel_type",
     "mpg", "mpg_high", "mpg_low", "mpg_avg"]
].sort_values("mpg_avg").head(50)
```

```python
df.loc[
    df["mpg_avg"] > 80,
    ["manufacturer", "model", "year", "fuel_type",
     "mpg", "mpg_high", "mpg_low", "mpg_avg"]
].sort_values("mpg_avg", ascending=False)
```

```python
df.loc[df["mpg_avg"] <= 0, ["mpg_high", "mpg_low", "mpg_avg"]] = np.nan
```

```python
df.loc[
    df["mpg_avg"] > 80,
    ["manufacturer", "model", "year", "fuel_type", "mpg", "mpg_avg"]
].groupby("fuel_type").size()
```

```python
df.loc[
    (df["mpg_avg"] > 80) & (df["fuel_type"].isna()),
    ["manufacturer", "model", "year", "fuel_type", "mpg", "mpg_avg"]
]
```

```python
# Treat zero/placeholder MPG values as missing
df.loc[df["mpg_avg"] <= 0, ["mpg_high", "mpg_low", "mpg_avg"]] = np.nan
```

```python
print("MPG <= 0:", (df["mpg_avg"] <= 0).sum())
print("Missing MPG:", df["mpg_avg"].isna().sum())
print("Maximum MPG:", df["mpg_avg"].max())
```

```python
df["engine_liters"] = (
    df["engine"]
    .str.extract(r"(\d+(?:\.\d+)?)L", expand=False)
    .astype(float)
)
```

```python
df[["engine", "engine_liters"]].head(20)
```

```python
df["engine_liters"].describe()
```

```python
df.loc[
    df["engine_liters"] <= 0,
    ["manufacturer", "model", "year", "engine", "engine_liters"]
].head(30)
```

```python
df.loc[
    df["engine_liters"] > 10,
    ["manufacturer", "model", "year", "engine", "engine_liters"]
].sort_values("engine_liters", ascending=False)
```

```python
df.loc[df["engine_liters"] <= 0, "engine_liters"] = np.nan
df.loc[df["engine_liters"] > 10, "engine_liters"] = np.nan
```

```python
df["engine_liters"].describe()
```

```python
df["is_electric"] = (
    df["fuel_type"]
    .str.contains("Electric", case=False, na=False)
    .astype(int)
)
```

```python
df["is_electric"].value_counts()
```

```python
categorical_cols = [
    "manufacturer",
    "transmission",
    "drivetrain",
    "fuel_type",
    "exterior_color",
    "interior_color"
]

for col in categorical_cols:
    print(f"\n--- {col} ---")
    print("Unique:", df[col].nunique(dropna=True))
    print(df[col].dropna().value_counts().head(10))
    
```

```python
for col in ["drivetrain", "fuel_type"]:
    print(f"\n--- {col} ---")
    print("Unique:", df[col].nunique(dropna=True))
    print(df[col].dropna().value_counts().head(20))
```

```python
print("\n--- transmission examples ---")
print(df["transmission"].dropna().value_counts().head(30))
```

```python
df["transmission_type"] = np.where(
    df["transmission"].str.contains("CVT", case=False, na=False),
    "CVT",
    np.where(
        df["transmission"].str.contains("Automatic", case=False, na=False),
        "Automatic",
        np.where(
            df["transmission"].str.contains("Manual", case=False, na=False),
            "Manual",
            "Other"
        )
    )
)
```

```python
df["transmission_gears"] = (
    df["transmission"]
    .str.extract(r"(\d+)-Speed", expand=False)
    .astype(float)
)
```

```python
df["is_automatic"] = (
    df["transmission_type"] == "Automatic"
).astype(int)
```

```python
print(df["transmission_type"].value_counts(dropna=False))
print()
print(df["transmission_gears"].describe())
```

```python
df["is_automatic"] = (
    df["transmission_type"].isin(["Automatic", "CVT"])
).astype(int)
```

```python
df["is_automatic"].value_counts()
```

```python
feature_cols = [
    "price",
    "year",
    "mileage",
    "vehicle_age",
    "engine_liters",
    "mpg_high",
    "mpg_low",
    "mpg_avg",
    "seller_rating",
    "driver_rating",
    "driver_reviews_num",
    "price_drop",
    "manufacturer",
    "fuel_type",
    "drivetrain",
    "transmission_type",
    "is_automatic",
    "is_electric",
    "has_accident",
    "is_one_owner",
    "is_personal_use"
]

missing_summary = pd.DataFrame({
    "missing_count": df[feature_cols].isna().sum(),
    "missing_pct": (df[feature_cols].isna().mean() * 100).round(2)
})

missing_summary.sort_values("missing_pct", ascending=False)
```

```python
df = df.dropna(subset=["price"]).copy()

print(f"Rows after removing missing target price: {len(df):,}")
print("Missing price:", df["price"].isna().sum())
```

```python
numeric_features = [
    "price",
    "year",
    "vehicle_age",
    "mileage",
    "engine_liters",
    "mpg_high",
    "mpg_low",
    "mpg_avg",
    "seller_rating",
    "driver_rating",
    "driver_reviews_num",
    "has_accident",
    "is_one_owner",
    "is_personal_use",
    "is_electric",
    "is_automatic",
    "transmission_gears"
]

corr = df[numeric_features].corr()

corr["price"].sort_values(ascending=False)
```

```python
output_path = "../data/processed/cars_preprocessed.csv"

df.to_csv(output_path, index=False)

print(f"Saved: {output_path}")
print(f"Rows: {len(df):,}")
print(f"Columns: {df.shape[1]}")
```
