```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/cars.csv")

print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]}")

df.head()
```

```python
df.info()
```

```python
# Missing values
missing = df.isnull().sum()

missing_percentage = (missing / len(df) * 100).round(2)

missing_report = pd.DataFrame({
    "missing_count": missing,
    "missing_percentage": missing_percentage
})

missing_report.sort_values("missing_count", ascending=False)
```

```python
# Duplicate rows
duplicate_count = df.duplicated().sum()


print(f"Duplicate rows: {duplicate_count:,}")
```

```python
# Statistical summary of numerical columns

df[["year", "mileage", "price", "price_drop"]].describe()
```

```python
# 10 most expensive cars
df.nlargest(10, "price")[[
    "manufacturer",
    "model",
    "year",
    "mileage",
    "price"
]]
```

```python

# 10 cars with the highest mileage
df.nlargest(10, "mileage")[[
    "manufacturer",
    "model",
    "year",
    "mileage",
    "price"
]]
```

```python
# Oldest 10 cars
df.nsmallest(10, "year")[[
    "manufacturer",
    "model",
    "year",
    "mileage",
    "price"
]]
```

```python
# Price distribution statistics
df["price"].describe(percentiles=[0.90, 0.95, 0.99, 0.995, 0.999])
```

```python
# Price distribution statistics
df["price"].describe(percentiles=[0.90, 0.95, 0.99, 0.995, 0.999])
```

```python
# Cars above $100,000
df[df["price"] > 100_000][[
    "manufacturer",
    "model",
    "year",
    "mileage",
    "price"
]].sort_values("price", ascending=False).head(30)
```

```python
# Cars with suspiciously low prices
df[df["price"] < 5_000][[
    "manufacturer",
    "model",
    "year",
    "mileage",
    "price"
]].sort_values("price").head(30)
```

```python
price_thresholds = [1, 100, 500, 1000, 5000, 10000, 100000, 500000, 1000000]

for threshold in price_thresholds:
    count = (df["price"] <= threshold).sum()
    print(f"Price <= ${threshold:,}: {count:,} rows")
```

```python
# Extremely high prices
print("Price > $500,000:", (df["price"] > 500_000).sum())
print("Price > $1,000,000:", (df["price"] > 1_000_000).sum())
print("Price > $5,000,000:", (df["price"] > 5_000_000).sum())
print("Price > $100,000,000:", (df["price"] > 100_000_000).sum())
```

```python
# Exact suspicious price values
df["price"].value_counts().head(30)
```

```python
high_price = df[df["price"] > 500_000][[
    "manufacturer",
    "model",
    "year",
    "mileage",
    "price"
]].sort_values("price", ascending=False)

high_price
```

```python
low_price = df[df["price"] < 1_000][[
    "manufacturer",
    "model",
    "year",
    "mileage",
    "price"
]].sort_values("price")

low_price
```

```python
df_clean = df.drop_duplicates().copy()
```

```python
df_clean = df.copy()

df_clean = df_clean.drop_duplicates()

df_clean.loc[df_clean["price"] > 100_000_000, "price"] = np.nan


df_clean.loc[df_clean["mileage"] >= 999_999, "mileage"] = np.nan

print(f"Original rows: {len(df):,}")
print(f"Cleaned rows: {len(df_clean):,}")
print(f"Rows removed: {len(df) - len(df_clean):,}")
```

```python
print("Duplicate rows:", df_clean.duplicated().sum())
print("Prices > $100M:", (df_clean["price"] > 100_000_000).sum())
print("Mileage >= 999,999:", (df_clean["mileage"] >= 999_999).sum())
```

```python
df_clean[["year", "mileage", "price"]].describe()
```

```python
import os

os.makedirs("../data/processed", exist_ok=True)

df_clean.to_csv(
    "../data/processed/cars_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
```
